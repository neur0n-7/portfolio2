* **Iteration (loops)**
  Using loops to continuously update objects like projectiles and to process collections of game objects. [Homework](https://neur0n-7.github.io/portfolio1/javascript/iteration/inventors-hw)

  ```js
  this.projectiles = this.projectiles.filter(p => !p.revComplete);
  this.projectiles.forEach(p => p.update());
  ```

  Looping through enemies to find the closest target:

  ```js
  for (const target of targets) {
      const dx = target.position.x - this.position.x;
      const dy = target.position.y - this.position.y;
      const dist = Math.sqrt(dx * dx + dy * dy);
  }
  ```

* **Conditionals (if/else)**
  Uses conditionals to control gameplay, such as preventing actions before the game starts or handling movement. [Homework](https://neur0n-7.github.io/portfolio1/csse/javascript/constellators/conditionals-hw)

  ```js
  if (!window.archeryGameStarted) {
      return;
  }
  ```

  ```js
  if (this.position.x <= 0) {
      this.velocity.x = this.speed;
  } else if (this.position.x + this.width >= this.gameEnv.innerWidth) {
      this.velocity.x = -this.speed;
  }
  ```

* **Nested Conditions (complex logic)**
  Combines multiple conditions to handle collision and win conditions. [Conditionals HW](https://neur0n-7.github.io/portfolio1/csse/javascript/constellators/conditionals-hw)

  ```js
  if (!this.stuck && xDiff <= TARGET_SIZE/2.0 && yDiff <= TARGET_SIZE/2.0) {
      this.stuck = true;
      nearestTarget.hitsRemaining -= 1;

      if (nearestTarget.hitsRemaining <= 0){
          showEndScreen(this.gameEnv);
      }
  }
  ```
